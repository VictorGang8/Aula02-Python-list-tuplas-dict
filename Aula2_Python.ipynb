**Lista: Uma lista é uma coleção de valores que pode mudar (você pode adicionar, remover,
alterar).**

```python
frutas = ["maça","banana","laranja"]


legumes = frutas
legumes.append("cenoura")
print(legumes)
```

Verificação em Listas/Tuplas

```python
frutas = ["maçã", "banana", "uva"]
print("banana" in frutas)  # Saída: True
print("laranja" in frutas) # Saída: False
```

**Tuplas: Parecido com lista, mas não pode mudar depois de criada (imutável).**

```python
coordenadas = (10,20,30,40,50)
print(coordenadas[4])
```

1-Crie um programa em Python que utilize listas, tuplas e dicionários para armazenar e exibir
informações de um aluno. O programa deve conter:

Uma lista chamada materias contendo três matérias que o aluno está cursando.

Uma tupla chamada notas contendo as notas do aluno em três provas.

Um dicionário chamado aluno contendo as seguintes informações:

*   Nome do aluno
*   Idade
*   Curso

O programa deve exibir a seguinte saída (os valores podem ser diferentes, desde que sigam a
estrutura):

Matérias: ['Matemática', 'História', 'Biologia']

Notas: (8.5, 7.0, 9.2)

Informações do aluno: {'nome': 'Carlos', 'idade': 21, 'curso': 'Engenharia'}

```python
MATERIAS = ["MATEMÁTICA","HISTÓRIA","BIOLOGIA"]
NOTAS = (8.5, 7.0, 9.2)
Alunos = {"nome": "Carlos", "idade": 21, "curso": "Engenharia"}
print("MATERIAS:", MATERIAS)
print("NOTAS:", NOTAS)
print("Informações do aluno:", Alunos)
```
